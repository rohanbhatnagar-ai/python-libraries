```python
# 6.
# Use map() with lambda to square every element of a list.
# Input:
# [1,2,3,4]

# Output:
# [1,4,9,16]

numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x * x, numbers))

print(squares)

```

```python
# 7.
# Use filter() with lambda to keep only even numbers.
# Input:
# [1,2,3,4,5,6]

# Output:
# [2,4,6]
numbers = [1,2,3,4,5,6]
even_number= list(filter(lambda x:x%2==0,numbers))
print(even_number)
```

```python
# 8.
# Use filter() with lambda to keep numbers greater than 50.
# Input:
# [20,45,55,70]

# Output:
# [55,70]
numbers =[20,45,55,70]
greater_number = list(filter(lambda x:x>=50,numbers))
print(greater_number)
```

```python
# 9.Sort a list of numbers in ascending order using lambda.
# [9,5,1,8,2]
numbers = [9,5,1,8,2]
as_number = sorted(numbers, key = lambda x:x)
print(as_number)
```

```python
# 10.
# Sort a list of numbers in descending order using lambda.
# [9,5,1,8,2]
numbers = [9,5,1,8,2]
as_number = sorted(numbers, key = lambda x:x,reverse=True)
print(as_number)
```

```python
#11.
# Use reduce() with lambda to find the sum of all numbers.
# [10,20,30,40]

from functools import reduce

numbers = [10, 20, 30, 40]

total = reduce(lambda x, y: x + y, numbers)

print(total)
```

```python

# Use reduce() with lambda to find the product of all numbers.
# [2,3,4,5]

numbers = [2, 3, 4, 5]

product = reduce(lambda x, y: x * y, numbers)

print(product)
```

```python
# 17.
# Use lambda with min() to find the employee with the lowest salary.
# employees = [
# ("A",35000),
# ("B",42000),
# ("C",28000)
# ]

employees = [
("A",35000),
("B",42000),
("C",28000)
]
lowest_salary=min(employees, key = lambda x:x)
print(lowest_salary)
```
