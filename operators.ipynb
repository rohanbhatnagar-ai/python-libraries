```python
# Question 7:
# Python
#  Take a word from user.
#  Check if the word is a palindrome or not.
#  (Example: "madam", "radar")

words = str(input("enter the word"))
if words ==words[::-1]:
    print("if the is a palindrome ")
else:
    print("not")
    
```
