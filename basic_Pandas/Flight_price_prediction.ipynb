```python
# Step 4: Data Preprocessing

# drop the missing values
df.dropna(inplace=True)

```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
# Drop the duplicated value 
df.drop_duplicates(inplace=True)
```

```python
# handled the first column Date_of_Journey

df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'],errors='coerce')

df['Day'] = df['Date_of_Journey'].dt.day
df['Month'] = df['Date_of_Journey'].dt.month
df['Year'] = df['Date_of_Journey'].dt.year
```

```python
#drop the Date_of_Journey column
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.columns
```

```python
#  drop Additional_Info
df.drop('Additional_Info',axis=1,inplace=True)
```

```python
df['Dep_Time']
```

```python
# extract the departure hour and minute from the Dep_Time column
df['Dep_Hour'] = pd.to_datetime(df['Dep_Time']).dt.hour
df['Dep_Minute'] = pd.to_datetime(df['Dep_Time']).dt.minute
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.columns
```

```python
# handle the Arrival_Time column
df['Arrival_Time'] = df['Arrival_Time'].str.split(' ',expand=True)[0]
df['Arrival_Time'] = pd.to_datetime(df['Arrival_Time'],errors='coerce')

df['Arrival_Hour'] = df['Arrival_Time'].dt.hour
df['Arrival_Minute'] = df['Arrival_Time'].dt.minute
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
# handled the Duration column
df['Duration'] = df['Duration'].str.replace('h','').str.replace('m','')
df['Duration_Hours'] = df['Duration'].str.split(' ',expand=True)[0]
df['Duration_Minutes'] = df['Duration'].str.split(' ',expand=True)[1]
```

```python
df['Duration_Minutes'].fillna(0,inplace=True)
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
df.head()
```

```python
# check the unique values in  the total_stops column
df['Total_Stops'].unique()
```

```python
#handled the Total_Stops column

df['Total_Stops'] = df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4})
```

```python
df.head()
```

```python
# handled the Route column

df['Route_1'] = df['Route'].str.split('→',expand=True)[0]
df['Route_2'] = df['Route'].str.split('→',expand=True)[1]
df['Route_3'] = df['Route'].str.split('→',expand=True)[2]
df['Route_4'] = df['Route'].str.split('→',expand=True)[3]
df['Route_5'] = df['Route'].str.split('→',expand=True)[4]
```

```python
df.drop('Route',axis=1,inplace=True)
```
