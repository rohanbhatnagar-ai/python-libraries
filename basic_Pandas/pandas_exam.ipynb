```python
# Part 8 — Final Real-World Cleaning Challenge
# Q29. Prepare the ML-ready dataset
# Perform the following complete pipeline:
# 1. Load the dataset.
# 2. Inspect the first 10 rows.
# 3. Inspect the last 10 rows.
# 4. Inspect 10 random records.
# 5. Run info().
# 6. Check missing values.
# 7. Check duplicates.
# 8. Remove duplicates.
# 9. Clean all numeric columns.
# 10. Convert invalid numeric values to missing.
# 11. Fill numerical missing values appropriately.
# 12. Convert Join_Date to datetime.
# 13. Handle invalid dates.
# 14. Standardize categorical values.
# 15. Clean the binary Active column.
# 16. Rename the columns.
# 17. Remove unnecessary columns.
# 18. Verify the final datatypes.
# 19. Generate descriptive statistics.
# 20. Check categorical distributions.

import pandas as pd

# 1. Load the dataset
df = pd.read_csv("C:\\Users\\ntt\\Downloads\\customer_analytics - customer_analytics (1).csv")
# 2. Inspect the first 10 rows
print(df.head(10))
# 3. Inspect the last 10 rows
print(df.tail(10))
# 4. Inspect 10 random records
print(df.sample(10))
# 5. Run info()
df.info()
# 6. Check missing values
print(df.isnull().sum())



# 7. Check duplicates
print( df.duplicated().sum())


# 8. Remove duplicates
df = df.drop_duplicates()

print(df.duplicated().sum())

# 9 & 10. CLEAN ALL NUMERIC COLUMNS

numeric_columns = [
    "Age",
    "Annual_Income",
    "Account_Balance",
    "Credit_Score",
    "Tenure_Years",
    "Monthly_Spend",
    "Transactions",
    "Satisfaction_Score"
]

# Convert invalid values to NaN
for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )


# 11. Fill numerical missing values

for column in numeric_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )

print(df[numeric_columns].isnull().sum())

# 12 & 13. CLEAN JOIN_DATE
# Convert Join_Date to datetime
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce"
)

print(df["Join_Date"].isnull().sum())


# Fill missing dates using median date
median_date = df["Join_Date"].median()

df["Join_Date"] = df["Join_Date"].fillna(
    median_date
)

print("Missing dates after cleaning:",
      df["Join_Date"].isnull().sum())

# 14. STANDARDIZE CATEGORICAL VALUE

# Clean City
df["City"] = (
    df["City"]
    .str.strip()
    .str.title()
)

print("\nCITY DISTRIBUTION")
print(df["City"].value_counts())


# Clean Segment
df["Segment"] = (
    df["Segment"]
    .str.strip()
)

# Fill missing Segment with most common category
df["Segment"] = df["Segment"].fillna(
    df["Segment"].mode()[0]
)
print(df["Segment"].value_counts())


# Clean Risk_Level
df["Risk_Level"] = (
    df["Risk_Level"]
    .str.strip()
    .str.title()
)

print("\nRISK LEVEL DISTRIBUTION")
print(df["Risk_Level"].value_counts())


# 15. CLEAN BINARY ACTIVE COLUMN


df["Active"] = (
    df["Active"]
    .astype(str)
    .str.strip()
)

df["Active"] = df["Active"].replace({
    "Y": "Yes",
    "YES": "Yes",
    "yes": "Yes",
    "1": "Yes",
    
    "N": "No",
    "NO": "No",
    "no": "No",
    "0": "No"
})

print("\nACTIVE DISTRIBUTION")
print(df["Active"].value_counts())

# 16. RENAME THE COLUMNS


df.rename(columns={
    "Customer_ID": "Customer_ID",
    "Annual_Income": "AnnualIncome",
    "Account_Balance": "AccountBalance",
    "Credit_Score": "CreditScore",
    "Tenure_Years": "Tenure",
    "Join_Date": "CustomerJoinDate",
    "Risk_Level": "RiskLevel",
    "Monthly_Spend": "MonthlySpend",
    "Satisfaction_Score": "SatisfactionScore"
}, inplace=True)


# Check renamed columns
print("\nRENAMED COLUMNS")
print(df.columns)

# 17. REMOVE UNNECESSARY COLUMN

# Customer_ID is an identifier, not a useful ML feature
df_clean = df.drop(
    columns=["Customer_ID"]
)
print(df_clean.head())

# 18. VERIFY FINAL DATATYPES
print(df_clean.dtypes)

# 19. GENERATE DESCRIPTIVE STATISTICS



print(df_clean.describe())

# 20. CHECK CATEGORICAL DISTRIBUTIONS



print(df_clean["City"].value_counts())


print("\nSEGMENT DISTRIBUTION")
print(df_clean["Segment"].value_counts())



print(df_clean["Active"].value_counts())

print(df_clean["RiskLevel"].value_counts())
```
