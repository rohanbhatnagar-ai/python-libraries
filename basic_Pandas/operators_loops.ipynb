```python
# 1. Take two numbers from user. Perform all arithmetic operations and also check
# which number is greater using comparison operators.
# Take two numbers from the user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Arithmetic Operations
print("\nArithmetic Operations:")
print("Addition =", num1 + num2)
print("Subtraction =", num1 - num2)
print("Multiplication =", num1 * num2)

if num2 != 0:
    print("Division =", num1 / num2)
    print("Floor Division =", num1 // num2)
    print("Modulus =", num1 % num2)
else:
    print("Division = Cannot divide by zero")
    print("Floor Division = Cannot divide by zero")
    print("Modulus = Cannot divide by zero")

print("Exponent =", num1 ** num2)

# Comparison Operators
print("\nComparison Results:")
print("num1 > num2 :", num1 > num2)
print("num1 < num2 :", num1 < num2)
print("num1 == num2 :", num1 == num2)
print("num1 != num2 :", num1 != num2)
print("num1 >= num2 :", num1 >= num2)
print("num1 <= num2 :", num1 <= num2)

# Check which number is greater
print("\nGreater Number:")
if num1 > num2:
    print(num1, "is greater than", num2)
elif num2 > num1:
    print(num2, "is greater than", num1)
else:
    print("Both numbers are equal")
```
