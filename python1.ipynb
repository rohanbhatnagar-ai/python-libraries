```python
# 41. Reverse a string without using slicing.
text = "python"
reverse = ""

for char in text:
    reverse = char + reverse
    print(reverse)
```

```python
numbers = [10, 20, 30, 20, 40, 10, 50]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print(duplicates)
```

```python
# 43. Find the first non-repeating character
text = "aabbde"

for char in text:
    if text.count(char) == 1:
        print(char)
        break
```

```python
# 44. Count frequency of elements.
numbers = [10, 20, 10, 30, 20, 10]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print(frequency)
```

```python
# 45. Find the second-largest number.
numbers = [10, 50, 30, 80, 40]

largest = numbers[0]
second_largest = numbers[0]

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print(second_largest)
```

```python
# 46. Find missing numbers from a sequence.
numbers = [1, 2, 4, 6, 7]

missing = []

for num in range(1, 8):
    if num not in numbers:
        missing.append(num)

print(missing)
```

```python
#47. Remove duplicates while preserving order
numbers = [10, 20, 10, 30, 20, 40, 30]

result = []

for num in numbers:
    if num not in result:
        result.append(num)

print(result)
```

```python
# 48. Flatten a nested list
nested = [[1, 2], [3, 4], [5, 6]]

result = []

for sublist in nested:
    for num in sublist:
        result.append(num)

print(result)
```

```python
# 49. Merge two dictionaries
dict1 = {"name": "Rohan", "age": 22}
dict2 = {"city": "Hyderabad", "course": "Python"}

result = {}

for key in dict1:
    result[key] = dict1[key]

for key in dict2:
    result[key] = dict2[key]

print(result)
```
