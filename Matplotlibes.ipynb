```python
# 1. Create a simple line plot using:
import matplotlib.pyplot as plt
import numpy as np
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]
plt.plot(X,Y)
plt.show()
```

```python
# 2. Create a line plot and add an appropriate title.
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]
plt.plot(X,Y)
plt.title("line plot")
plt.show()
```

```python
# 3. Create a line plot and label both the X-axis and Y-axis

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x,y)
plt.title("line plot")
plt.xlabel("X-axises")
plt.ylabel("Y-sxises")
plt.show()
```

```python
# 4. Plot two lines on the same graph and add a legend to identify them.

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x,y)
plt.title("line plot")
plt.xlabel("X-axises")
plt.ylabel("Y-sxises")

plt.legend("two line graph")
plt.show()
```

```python
# 5. Change the line color, line style, and marker style of a line plots

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x,y,
        color = "blue",
        linewidth = 3,
        markersize =8,
        markeredgecolor = "green",
        markerfacecolor = "white",
        
        )


plt.title("line plot")
plt.xlabel("X-axises")
plt.ylabel("Y-sxises")

plt.legend("two line graph")
plt.show()
```

```python
# 6. Create a line plot using NumPy values generated from np.arange().
import numpy as np
x = np.linspace(0,20,100)

y = np.sin(x)

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x,y,
        color = "blue",
        linewidth = 3,
         
        markersize =8,
        markeredgecolor = "green",
        markerfacecolor = "white",
        
        )
plt.title("line plot")
plt.xlabel("X-axises")
plt.ylabel("Y-sxises")

plt.legend("two line graph")
plt.show()
```

```python
# 7. Create a bar chart showing the sales of 5 different products.
product = ["laptop","tv","mobile","watch","macpro"]
sales=[70 , 76, 76, 45, 67]
plt.bar(product,sales)
plt.xlabel("sales")
plt.ylabel("prodcut")
plt.legend("bar chart")
plt.show()
```

```python
# 8. Create a horizontal bar chart for the same product-sales data.
product = ["laptop","tv","mobile","watch","macpro"]
sales=[70 , 76, 76, 45, 67]
plt.barh(product,sales)
plt.xlabel("sales")
plt.ylabel("prodcut")
plt.legend("bar chart")
plt.show()
```

```python
# 9. Create a scatter plot between Age and Salary for 10 employees.
age = [23,45,56,23,45,5,56]
salary = [65000,75000,560000,450000,66000,670000,78000]
plt.scatter(age,salary)
plt.xlabel("employee salary")
plt.ylabel("employee age")
plt.show()
```

```python
# 10.Create a histogram showing the distribution of 100 randomly generated
# values
import numpy as np
data =  np.random.randn(100)
plt.hist(data)
plt.title("distribution of 100 randomly generated")
plt.xlabel("values")
plt.ylabel("sales")
plt.show()
```

```python
# 11.Create a pie chart showing the percentage of customers from 4 different
# cities.
cities = ["Bangalore","hyderabad","indore","pune"]
customers = [10, 35,67,86]
plt.pie(customers , labels=cities, autopct="%1.1f%%")

plt.xlabel("customers")
plt.ylabel("cities")
plt.legend("+")
plt.show()
```
