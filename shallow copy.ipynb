```python
# 7. Write a function manual_deep_copy(lst) that manually creates a deep copy of a nested list
# without using copy.deepcopy().
def manual_deep_copy(lst):
    new_list = []

    for item in lst:
        if isinstance(item, list):
            new_list.append(manual_deep_copy(item))
        else:
            new_list.append(item)

    return new_list


# Original nested list
original = [[1, 2], [3, [4, 5]], 6]

# Manual deep copy
copied = manual_deep_copy(original)

# Modify the copied list
copied[1][1][0] = 100

print("Original List:", original)
print("Copied List:", copied)
```

```python
# 8. Given this data: data = [10, [20, 30], {"name": "Ram"}, 40] Make shallow and deep copies.
# Then change data[1][0] and data[2]["name"]. Print all three versions.
import copy

# Original data
data = [10, [20, 30], {"name": "Ram"}, 40]

# Shallow copy
shallow = copy.copy(data)

# Deep copy
deep = copy.deepcopy(data)

# Modify the original data
data[1][0] = 100
data[2]["name"] = "Rohan"

# Print all three versions
print("Original Data :", data)
print("Shallow Copy  :", shallow)
print("Deep Copy     :", deep)
```

```python
# 9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a
# nested element and checking the original.
import copy

def detect_copy(original, copied):
    # Save original value
    old_value = original[1][0]

    # Modify the nested element in the copied list
    copied[1][0] = 999

    # Check if original changed
    if original[1][0] == 999:
        print("This is a Shallow Copy")
    else:
        print("This is a Deep Copy")

    # Restore original value
    original[1][0] = old_value


# Original list
data = [10, [20, 30], 40]

# Shallow copy
shallow = copy.copy(data)

# Deep copy
deep = copy.deepcopy(data)

print("Checking the shallow copy:")
detect_copy(data, shallow)

print("\nChecking deep copy:")
detect_copy(data, deep)
```
