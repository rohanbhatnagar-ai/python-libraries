```python
# 1. Create a class Student with instance attributes name and age. Create two objects
# and print their details.

class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display(self):
        print("name:",self.name)
        print("age:",self.age)
        print()
# object
student1=Student("rohan",20)
student2=Student("vadika",24)
student1.display()
student2.display()
```

```python
# 2. Define class Car with attributes brand, model, and year. Create an object and
# display all attributes.


class Car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def display(self):
        print("brand:",self.brand)
        print("model:",self.model)
        print()
# object
car1=Car("bmw","l2")
car2=Car("Mercedes","2012")
car1.display()
car2.display()
```

```python
# 3. Create class Book with attributes title, author, price. Add a method display_info().


class Book:
    def __init__(self,title,author,prices):
        self.title=title
        self.author=author
        self.prices=prices
    def display_info(self):
        print("Title:",self.title)
        print("Author:",self.author)
        print("Prices:",self.prices)
# object
book1 = Book("the wild ","james",10000)
book2 = Book("the adventure","brook",30000)
book1.display_info()
book2.display_info()
```
