```python
# # 8. Local vs Global Variable
# Python
# # Create a global variable 'college'
# # Inside a function, try to print it (use global keyword if needed)
# # Demonstrate local variable inside function
# Global variable
college = "ABC College"

def student_details():
    # Accessing global variable
    global college
    print("College:", college)

    # Local variable
    student_name = "Rohan"
    print("Student Name:", student_name)

# Call the function
student_details()

# Global variable can be accessed outside
print("Outside function:", college)
```

```python
# Medium Level Practicals
# 9. Simple Calculator
# Python
# # Create a simple calculator that takes two numbers and one operator (+,-,*,/,//,%,**)
# # from the user and shows the result.
# Take two numbers from the user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Take operator from the user
operator = input("Enter operator (+, -, *, /, //, %, **): ")

# Perform calculation
if operator == "+":
    print("Result:", num1 + num2)

elif operator == "-":
    print("Result:", num1 - num2)

elif operator == "*":
    print("Result:", num1 * num2)

elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Cannot divide by zero")

elif operator == "//":
    if num2 != 0:
        print("Result:", num1 // num2)
    else:
        print("Cannot divide by zero")

elif operator == "%":
    if num2 != 0:
        print("Result:", num1 % num2)
    else:
        print("Cannot divide by zero")

elif operator == "**":
    print("Result:", num1 ** num2)

else:
    print("Invalid operator")
```

```python
# 10. Temperature Converter
# Python

# Sumit Sir Notes
# # Convert temperature from Celsius to Fahrenheit and vice versa.
# # Formula: F = (C * 9/5) + 32
print("Temperature Converter")

choice = input("Enter C for Celsius to Fahrenheit or F for Fahrenheit to Celsius: ")

if choice == "C" or choice == "c":
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = (celsius * 9/5) + 32
    print("Temperature in Fahrenheit:", fahrenheit)

elif choice == "F" or choice == "f":
    fahrenheit = float(input("Enter temperature in Fahrenheit: "))
    celsius = (fahrenheit - 32) * 5/9
    print("Temperature in Celsius:", celsius)

else:
    print("Invalid choice")
```

```python
# 11. Area and Perimeter Calculator
# Python
# # Take radius of circle from user and calculate:
# # Area and Circumference of circle
# # (Use pi = 3.14)
# Take radius from user
radius = float(input("Enter radius of circle: "))

pi = 3.14

# Calculate area
area = pi * radius ** 2

# Calculate circumference
circumference = 2 * pi * radius

# Display results
print("Area of circle:", area)
print("Circumference of circle:", circumference)
```

```python
# 12. Student Grade Calculator
# Python
# # Take marks of 5 subjects.
# # Calculate percentage and print grade:
# # >=90 → A+, >=80 → A, >=70 → B, >=60 → C, else → Fail
# Take marks of 5 subjectes

sub1 = float(input("Enter marks of Subject 1: "))
sub2 = float(input("Enter marks of Subject 2: "))
sub3 = float(input("Enter marks of Subject 3: "))
sub4 = float(input("Enter marks of Subject 4: "))
sub5 = float(input("Enter marks of Subject 5: "))

# Calculate total and percentages
total = sub1 + sub2 + sub3 + sub4 + sub5
percentage = total / 5

# Calculate grade
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
else:
    grade = "Fail"

# Display results
print("Total Marks:", total)
print("Percentage:", percentage, "%")
print("Grade:", grade)
```

```python
# 13. Swap Two Variables
# Python
# # Take two numbers from user and swap them using:
# # a) Third variable
# # b) Without third variable (Python way)
# Take two numbers from user
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Before swapping:")
print("a =", a)
print("b =", b)

# Swap using third variable
temp = a
a = b
b = temp

print("After swapping:")
print("a =", a)
print("b =", b)
```
