```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# pyplot : basic plot 

X = [1,2,3,4,5]

Y = [10,11,12,13,14]

plt.plot(X,Y)
plt.show
```

```python
#  basic plot with label

plt.plot(X,Y,
         color = "Red",
        linewidth=0.5,
        marker = "*",
        markersize = 8,
        markeredgecolor='blue',
        markerfacecolor='green',
        )

plt.title("First Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
# Pyplot Common Commands 

X = np.linspace(0,10,100)

y = np.sin(X)


# create the plot
plt.plot(X,y,
         color = "Red",
        linewidth=0.5,
        marker = "*",
        markersize = 8,
        markeredgecolor='purple',
        markerfacecolor='black',
        )

# commands
plt.title("Sine Waves")
plt.xlabel("X Axis Data ")
plt.ylabel("Sin X Data ")
plt.xlim(1,10)      # xrange
plt.ylim(-1.5, 1.5)  # yrange
plt.xticks([0,2,4,6,8,10])      # set x tick marks
plt.yticks([-1, 0, 1])          # set y tick marks 
plt.grid(True, alpha=0.9)       # show grid 
plt.legend(["Sin X Waves"])          # add legend 
plt.tight_layout()              # auto-fix spacing
plt.show()
```

# Figure Class

```python
#create the figure

plt.figure(figsize = (10,6),  # width=10, height=6 inches 
         dpi =50,             # dots per inch (quality) 
        facecolor = "yellow",  # background color
        edgecolor = "red" )        




plt.plot(X,y)
plt.show
```

```python
# figure methods


#create the figure

fig=plt.figure(figsize = (8,6))  


# figure method
fig.set_size_inches(5,4)   # resize figure in inches
fig.set_facecolor("lightblue")  # change background color
fig.suptitle("Figure Title",
              fontsize=26,  # font size figure title
             fontweight='bold')
 

# save the figure
fig.savefig("my_plot.png",
           dpi = 60,
           bbox_inches='tight')

# make plot 
plt.plot(X,y)

plt.show()
```

```python
# Multiplot  figure 

X =  np.linspace(0,10,100)

#create the multiplot figure

fig,axes = plt.subplots(2,2,
                        dpi = 60,
                       figsize = (12,8))


# figure method

fig.set_facecolor("lightgreen")  # change background color
fig.suptitle("Figure Title",
              fontsize=26,  # font size figure title
             fontweight='bold')
 


# Plot in each subplot 
axes[0,0].plot(X, np.sin(X), 'olive') 
axes[0,0].set_title("Sine plot") 

# Plot in each subplot 
axes[0,1].plot(X, np.cos(X), 'slategray') 
axes[0,1].set_title("Cos plot") 

# Plot in each subplot 
axes[1,0].plot(X, np.tanh(X), 'orange') 
axes[1,0].set_title("Tanh plot") 

# Plot in each subplot 
axes[1,1].plot(X, X**2, 'tan') 
axes[1,1].set_title("Squred plot") 

plt.show()
```

```python
# Multiplot  figure with Axes methods 

X =  np.linspace(0,10,100)

#create the multiplot figure

fig,axes = plt.subplots(2,2,
                        dpi = 60,
                       figsize = (15,12))


# figure method

fig.set_facecolor("lightgreen")  # change background color
fig.suptitle("Figure Title",
              fontsize=26,  # font size figure title
             fontweight='bold')
 


# Plot in each subplot 
axes[0,0].plot(X, np.sin(X), 'olive', linewidth=4) 
axes[0,0].set_title("Sine plot",fontsize=16)
axes[0,0].set_xlabel("X Axis Label", fontsize=12) 
axes[0,0].set_ylabel("Y Axis Label", fontsize=12) 
axes[0,0].set_xlim(0,12)
axes[0,0].set_ylim(-1.5,1.5)
axes[0,0].legend(["Sine Waves"])
axes[0,0].grid(True, linestyle='--', alpha=0.9) 
axes[0,0].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')


# Plot in each subplot 
axes[0,1].plot(X, np.cos(X), 'slategray',linewidth=4) 
axes[0,1].set_title("Cos plot",fontsize=16) 
axes[0,1].set_xlabel("X Axis Label", fontsize=12) 
axes[0,1].set_ylabel("Y Axis Label", fontsize=12)
axes[0,1].set_xlim(0,12)
axes[0,1].set_ylim(-1.5,1.5)
axes[0,1].legend(["Cose Waves"])
axes[0,1].grid(True, linestyle='--', alpha=0.9) 
axes[0,1].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')

# Plot in each subplot 
axes[1,0].plot(X, np.tanh(X), 'orange',linewidth=4) 
axes[1,0].set_title("Tanh plot",fontsize=16) 
axes[1,0].set_xlabel("X Axis Label", fontsize=12)
axes[1,0].set_ylabel("Y Axis Label", fontsize=12)
axes[1,0].set_xlim(0,10)
axes[1,0].set_ylim(0,1.25)
axes[1,0].legend(["Tanh Waves"])
axes[1,0].grid(True, linestyle='--', alpha=0.9) 
axes[0,1].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')

# Plot in each subplot 
axes[1,1].plot(X, X**2, 'tan',linewidth=4) 
axes[1,1].set_title("Squred plot",fontsize=16)
axes[1,1].set_xlabel("X Axis Label", fontsize=12) 
axes[1,1].set_ylabel("Y Axis Label", fontsize=12)
axes[1,1].set_xlim(0,10)
axes[1,1].set_ylim(0,100)
axes[1,1].legend(["Squared Waves"])
axes[1,1].grid(True, linestyle='--', alpha=0.9) 


plt.show()
```

```python
# Multiplot  figure with Axes methods 

X =  np.linspace(0,10,100)

#create the multiplot figure

fig,axes = plt.subplots(2,3,
                        dpi = 60,
                       figsize = (15,12))


# figure method

fig.set_facecolor("lightgreen")  # change background color
fig.suptitle("Figure Title",
              fontsize=26,  # font size figure title
             fontweight='bold')
 


# Plot in each subplot 
axes[0,0].plot(X, np.sin(X), 'olive', linewidth=4) 
axes[0,0].set_title("Sine plot",fontsize=16)
axes[0,0].set_xlabel("X Axis Label", fontsize=12) 
axes[0,0].set_ylabel("Y Axis Label", fontsize=12) 
axes[0,0].set_xlim(0,12)
axes[0,0].set_ylim(-1.5,1.5)
axes[0,0].legend(["Sine Waves"])
axes[0,0].grid(True, linestyle='--', alpha=0.9) 
axes[0,0].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')


# Plot in each subplot 
axes[0,1].plot(X, np.cos(X), 'slategray',linewidth=4) 
axes[0,1].set_title("Cos plot",fontsize=16) 
axes[0,1].set_xlabel("X Axis Label", fontsize=12) 
axes[0,1].set_ylabel("Y Axis Label", fontsize=12)
axes[0,1].set_xlim(0,12)
axes[0,1].set_ylim(-1.5,1.5)
axes[0,1].legend(["Cose Waves"])
axes[0,1].grid(True, linestyle='--', alpha=0.9) 
axes[0,1].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')

# Plot in each subplot 
axes[0,2].bar([1,2,3,4,6], [4,5,6,2,7], color='green')  
axes[0,2].set_title("Bar plot",fontsize=16) 
axes[0,2].set_xlabel("X Axis Label", fontsize=12) 
axes[0,2].set_ylabel("Y Axis Label", fontsize=12)
axes[0,2].set_xlim(0,8)
axes[0,2].legend(["Cose Waves"])
axes[0,2].grid(True, linestyle='--', alpha=0.9) 



# Plot in each subplot 
axes[1,0].plot(X, np.tanh(X), 'orange',linewidth=4) 
axes[1,0].set_title("Tanh plot",fontsize=16) 
axes[1,0].set_xlabel("X Axis Label", fontsize=12)
axes[1,0].set_ylabel("Y Axis Label", fontsize=12)
axes[1,0].set_xlim(0,10)
axes[1,0].set_ylim(0,1.25)
axes[1,0].legend(["Tanh Waves"])
axes[1,0].grid(True, linestyle='--', alpha=0.9) 
axes[1,0].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')

# Plot in each subplot 
axes[1,1].plot(X, X**2, 'tan',linewidth=4) 
axes[1,1].set_title("Squred plot",fontsize=16)
axes[1,1].set_xlabel("X Axis Label", fontsize=12) 
axes[1,1].set_ylabel("Y Axis Label", fontsize=12)
axes[1,1].set_xlim(0,10)
axes[1,1].set_ylim(0,100)
axes[1,1].legend(["Squared Waves"])
axes[1,1].grid(True, linestyle='--', alpha=0.9) 


# Plot in each subplot 
axes[1,2].scatter(X[:50], np.sin(X[:50]),color='purple', s=20) 
axes[1,2].set_title("Squred plot",fontsize=16)
axes[1,2].set_xlabel("X Axis Label", fontsize=12) 
axes[1,2].set_ylabel("Y Axis Label", fontsize=12)
axes[1,2].set_xlim(0,5)
axes[1,2].set_ylim(0,1)
axes[1,2].legend(["Squared Waves"])
axes[1,2].grid(True, linestyle='--', alpha=0.9) 

plt.show()
```
