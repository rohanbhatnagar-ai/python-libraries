```python
# 16. Use while loop to reverse a number (e.g., 1234 → 4321).
num1 = int(input("enter reverse"))
reverse = 0
while num1>=0:
    digit = num1*10
    reverse = reverse*10+digit
    num1 = num1//10
    print("reverse",reverse)
```
