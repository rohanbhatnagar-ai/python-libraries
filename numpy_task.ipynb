```python
# 10.Find all unique values in the customer_segment column and identify
# inconsistent categories such as Consumer, consumer, consumer , etc.
df["customer_segment"].unique()
```

```python
# 11.Clean the customer_segment column by standardizing capitalization and
# whitespace.
df["customer_segment"] = df["customer_segment"].astype(str).str.rstrip().str.title()
print(df["customer_segment"])
```

```python
# 12.Find all unique values in the gender column and convert them into consistent
# categories such as Male and Female.
df["gender"] = df["gender"].astype(str).str.strip().str.lower()
df["gender"] = df["gender"].replace({
    "nan":"Male",
    "male":"Male",
    "m":"Male",
    "m":"Female",
    "m":"Female",
    "f":"Female",})
print(df["gender"].unique())
```

```python
# 13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and
# No, no, N, 0 become 0.
df["is_active"] = df["is_active"].astype(str).str.strip().str.lower()
print(df["is_active"])
df["is_active"].replace({
    "yes":1,
    "yes":1,
    "n":1,
    "No":0,
    "n":0,
    "no":0,})
print(df["is_active"])
    
```

```python
# 14.Convert the churn_flag column from categorical values (Yes/No) into a binary
# numerical column (1/0).
df["churn_flag"] = df["churn_flag"].astype(str).str.strip().str.lower()
df["churn_flag"]
df["churn_flag"].replace({
    "yes":1,
    "no":0
})
df["churn_flag"] = pd.to_numeric(df["churn_flag"], errors="coerce")
print(df["churn_flag"].unique())
```
