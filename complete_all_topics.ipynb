```python
# 2. Print all even numbers from 1 to 50 using for loop and skip numbers divisible by 10 using
# continue.
for i in range(1,51):
    if i%10==0:
        continue
    if i%2==0:
        print(i)
```

```python
# 3. Take numbers from user until they enter 0. Calculate sum using while loop.
total = 0 
while True:
    nums =int(input("enter the number (0 to stop:)"))
    if nums == 0:
        break
    total = total + nums
print("total numbers:",total)
```
